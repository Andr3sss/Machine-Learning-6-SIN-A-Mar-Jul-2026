```python
#COMMIT 1

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the default figure size for the plots in this notebook.
plt.rcParams['figure.figsize'] = (14, 7)
plt.style.use('fivethirtyeight')

# Import display so we can show tables and results nicely in the notebook.
from IPython.display import display
```

```python
#COMMIT 2

df = pd.read_csv('penguins_size.csv')

display(df.head())
print(f'Rows: {df.shape[0]}')
print(f'Columns: {df.shape[1]}')
```

```python
#COMMIT 3

# Count missing values in each column.
null_counts = df.isna().sum()
display(null_counts.to_frame(name='null_count'))
```

```python
#COMMIT 4

# Calculate the percentage of missing values in each column.
null_percentages = df.isna().mean() * 100
display(null_percentages.to_frame(name='null_percentage'))
print(f"Max null percentage: {null_percentages.max():.2f}%")
```

```python
#COMMIT 5

df = df.dropna()
df['sex'].unique()
```

```python
#COMMIT 6

df = df.dropna()
df = df[~df['sex'].astype(str).str.contains(r'\.', regex=True, na=False)]
display(df['sex'].unique())
```

```python
pd.get_dummies(df)
```

```python
pd.get_dummies(df.drop('species',axis=1),drop_first=True)
```

This code converts the categorical variables in the dataframe into binary numeric variables.

In short, it is used to prepare the data before training a machine learning model.

```python
X = pd.get_dummies(df.drop('species',axis=1),drop_first=True)
y = df['species']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=483)
```

# Decision Tree Classifier
## Default Hyperparameters

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)

base_pred = model.predict(X_test) 
```

# EVALUATION

```python
from sklearn.metrics import confusion_matrix,classification_report,plot_confusion_matrix
confusion_matrix(y_test,base_pred)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, base_pred)
ConfusionMatrixDisplay(cm).plot()
```

```python
print(classification_report(y_test,base_pred))
```

**Support** es la cantidad de muestras reales que tiene cada clase en el conjunto de prueba.

En otras palabras, indica cuántas veces aparece cada etiqueta en `y_test`.  
Sirve para saber si una clase está más representada que otra y para interpretar mejor las métricas del reporte.

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12,8))
plot_tree(model);
```

```python
plt.figure(figsize=(12,8),dpi=150)
plot_tree(model,filled=True,feature_names=X.columns);
```
